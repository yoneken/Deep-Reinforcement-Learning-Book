```python
import sys, os, time
from xarm.wrapper import XArmAPI
```

## テスト

```python
arm = XArmAPI("192.168.0.244", is_radian=True)

if arm.warn_code != 0:
    arm.clean_warn()
if arm.error_code != 0:
    arm.clean_error()
    
arm.motion_enable(enable=True)
arm.set_mode(7) # online cartesian mode
arm.set_state(state=0)
```

```python
arm.set_position(x=360, y=0, z=370, roll=120, pitch=-90, yaw=60, speed=400, is_radian=False, wait=False)
print(arm.get_position(), arm.get_position(is_radian=False))
time.sleep(0.5)
arm.set_position(x=360, y=240, z=370, roll=120, pitch=-90, yaw=60, speed=400, is_radian=False, wait=False)
print(arm.get_position(), arm.get_position(is_radian=False))
time.sleep(0.5)
arm.set_position(x=360, y=-240, z=370, roll=120, pitch=-90, yaw=60, speed=400, is_radian=False, wait=False)
print(arm.get_position(), arm.get_position(is_radian=False))
time.sleep(0.5)
arm.set_position(x=360, y=240, z=370, roll=120, pitch=-90, yaw=60, speed=400, is_radian=False, wait=False)
print(arm.get_position(), arm.get_position(is_radian=False))
time.sleep(0.5)
arm.set_position(x=360, y=-240, z=370, roll=120, pitch=-90, yaw=60, speed=400, is_radian=False, wait=False)
print(arm.get_position(), arm.get_position(is_radian=False))
time.sleep(0.5)
arm.set_position(x=360, y=240, z=370, roll=120, pitch=-90, yaw=60, speed=400, is_radian=False, wait=False)
print(arm.get_position(), arm.get_position(is_radian=False))
time.sleep(0.5)
arm.set_position(x=360, y=-240, z=370, roll=120, pitch=-90, yaw=60, speed=400, is_radian=False, wait=False)
print(arm.get_position(), arm.get_position(is_radian=False))

arm.disconnect()
```

```python
# Move to Y min
arm.set_position(x=360, y=240, z=370, roll=120, pitch=-90, yaw=60, speed=400, is_radian=False, wait=True)
print(arm.get_position(), arm.get_position(is_radian=False))

# realsense: x=0.235m, y=0.042m, z=0.554m
```

```python
# Move to Y max
arm.set_position(x=360, y=-240, z=370, roll=120, pitch=-90, yaw=60, speed=400, is_radian=False, wait=True)
print(arm.get_position(), arm.get_position(is_radian=False))

# realsense: x=-0.232m, y=0.052m, z=0.547m
```
